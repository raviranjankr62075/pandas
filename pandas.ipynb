```python
import numpy as np 
import pandas as pd 
```

```python
df = pd.read_csv("anime.csv")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df['Anime_Name'] = df['Title'].str.extract(r'^(.*?)(?:TV|OVA|Movie|Special)')
df['Episodes'] = df['Title'].str.extract(r'\((\d+)\s*eps\)').astype(int)
df['Total Time'] = df['Title'].str.extract(r'\((?:\d+)\s*eps\)([A-Za-z]{3}\s*\d{4}\s*-\s*[A-Za-z]{3}\s*\d{4})')
time_split = df['Total Time'].str.split(' - ', expand=True)
start_dates = pd.to_datetime(time_split[0], format='%b %Y')
end_dates = pd.to_datetime(time_split[1], format='%b %Y')
df['Months'] = (end_dates.dt.year - start_dates.dt.year) * 12 + (end_dates.dt.month - start_dates.dt.month) + 1
df.head()
```

```python
# Question 3: Which anime has the highest score
highest_score_anime = df[df['Score'] == df['Score'].max()]
print("Highest Scoring Anime:")
print(highest_score_anime[['Rank', 'Anime_Name', 'Score', 'Episodes', 'Months']])
```

```python
# Question 4: Give me top 5 highest scoring anime
top_5_score = df.sort_values(by='Score', ascending=False).head(5)
print("Top 5 Highest Scoring Anime:")
print(top_5_score[['Rank', 'Anime_Name', 'Score', 'Episodes', 'Months']])
```

```python
# Question 5: Which anime has the highest episode count
max_eps_anime = df[df['Episodes'] == df['Episodes'].max()]
print("Highest Episode Count Anime:")
print(max_eps_anime[['Rank', 'Anime_Name', 'Episodes', 'Score', 'Months']])
```

```python
# Question 6: Animes with top 5 episode count
top_5_eps = df.sort_values(by='Episodes', ascending=False).head(5)
print("Top 5 Anime by Episode Count:")
print(top_5_eps[['Rank', 'Anime_Name', 'Episodes', 'Score', 'Months']])
```

```python
# Question 7: Which is the longest running anime
longest_anime = df[df['Months'] == df['Months'].max()]
print("Longest Running Anime:")
print(longest_anime[['Rank', 'Anime_Name', 'Months', 'Total Time', 'Episodes', 'Score']])
```
